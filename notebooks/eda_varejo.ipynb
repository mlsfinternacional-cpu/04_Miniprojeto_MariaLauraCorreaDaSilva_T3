### Dicionário das variáveis

| Coluna | Descrição |
|---|---|
| `DATA` | Data da compra |

| `CL_ID` | Identificação do cliente |

| `PR_NOME` | Nome do produto adquirido |

### Agrupamento 1- Qual gênero realizou mais compras?

````python
df.groupby("CL_GENERO")["CO_ID"].nunique()
````

````python
df.groupby("CL_GENERO")["CO_ID"].nunique().sort_values().plot(
    kind="bar",
    title="Compras distintas por gênero",
    xlabel="Gênero",
    ylabel="Quantidade de compras",
)
````

### Interpretação

O agrupamento mostra que o sexo biológico informado como `F` apresentou 9.615 compras distintas, enquanto `M` apresentou 8.856. Portanto, `F` realizou o maior número de compras na base analisada.

### Agrupamento 2- Qual categoria de produto apresenta maior número de compras?

````python
df.groupby("PR_CAT")["CO_ID"].nunique().sort_values(ascending=False)
````

````python
df.groupby("PR_CAT")["CO_ID"].nunique().sort_values(ascending=True).plot(
    kind="barh",
    title="Compras distintas por categoria de produto",
    xlabel="Quantidade de compras",
    ylabel="Categoria do produto",
)
````

### Interpretação

O agrupamento mostra que a categoria `ALIMENTOS` apresentou o maior número de compras distintas, com 18.267 compras. Em seguida aparecem `HIGIENE`, com 17.610, e `LIMPEZA`, com 17.475. A categoria `Sem Categoria` apresentou o menor número de compras distintas, com 3.228 registros.

## Conclusão do Sprint 5

Os agrupamentos realizados permitiram identificar padrões no comportamento de compras da base. A análise por sexo informado mostrou maior número de compras distintas para `F`, com 9.615 compras, contra 8.856 para `M`.

Na análise por categoria de produto, `ALIMENTOS` apresentou o maior número de compras distintas, com 18.267, seguida por `HIGIENE` e `LIMPEZA`. A categoria `Sem Categoria` apresentou o menor volume entre as categorias analisadas.

Os resultados permitiram responder às perguntas propostas e evidenciar diferenças no volume de compras entre os grupos analisados.
