### 3.1- Verificando nulos por coluna

````python
df.isnull().sum()
````

### Identificando quais colunas apresentam problemas.

````python
nulos = df.isnull().sum()
nulos[nulos > 0]
````

### Tratamento das colunas vazias: As únicas colunas com problemas são Unnamed: 10 a Unnamed: 13, todas com 830.000 valores nulos, ou seja, 100% da coluna. Como não carregam informação observável, o tratamento adequado é removê-las da base.

### Remoção das colunas

````python
df = df.drop(columns=nulos[nulos == len(df)].index)
````

### Validação

````python
df.shape
````

````python
df.isnull().sum()
````

### 3.2- Verificando Duplicatas

````python
df.duplicated().sum()
````

### Avaliação das Duplicatas

A verificação identificou 95.553 registros duplicados na base.
Antes de removê-los, é necessário avaliar se essas repetições representam
registros indevidos ou se correspondem à estrutura natural dos dados.

````python
df[df.duplicated(keep=False)].head(10)
````

### Identificando a natureza das Duplicatas

````python
df[df.duplicated(keep=False)].sort_values(by=["DATA", "CO_ID", "PR_ID"]).head(20)
````

````python
df[df.duplicated(keep=False)].value_counts().head(10)
````

### Tratamento das Duplicatas

A análise confirmou que os registros identificados como duplicados são
cópias exatas de linhas já presentes na base, com algumas ocorrências
repetidas múltiplas vezes.

Como essas repetições não acrescentam informação à análise, será mantida
uma única ocorrência de cada registro e as duplicatas excedentes serão
removidas.

````python
df = df.drop_duplicates()
````

### Validação

````python
df.shape
````

````python
df.duplicated().sum()
````

### 3.3- Verificando as Categorias

A coluna `PR_CAT` será analisada para identificar categorias vazias ou
inconsistentes antes da definição do tratamento.

````python
df["PR_CAT"].value_counts(dropna=False)
````

### Isolando #N/D

````python
df[df["PR_CAT"] == "#N/D"].head(10)
````

````python
(df["PR_CAT"] == "#N/D").sum()
````

### Tratamento das Categorias

A análise identificou 3.228 registros com `#N/D` na coluna `PR_CAT`.
A inspeção desses registros mostrou que `PR_NOME` também está como `#N/D`,
indicando ausência de informação de categoria na fonte.

Como esses registros não possuem uma categoria observável, o valor será
padronizado como `Sem Categoria`, preservando os registros sem atribuir
uma classificação que não está presente na fonte.

````python
df["PR_CAT"] = df["PR_CAT"].replace("#N/D", "Sem Categoria")
````

### Validação

````python
df["PR_CAT"].value_counts(dropna=False)
````

````python
(df["PR_CAT"] == "#N/D").sum()
````
